###Q.2 Write a code to count the number of vowels in a string.

```python
def count_vowels(text):
    vowels = "aeiou"
    count = 0

    for char in text.lower():
        if char in vowels:
            count += 1

    return count

# Example
print(count_vowels("Hello World"))  # Output: 3
```

###Q.5 Write a code to find all occurrences of a given substring within another string.

```python
def find_substring_occurrences(text, substring):
    positions = []
    start = 0

    while True:
        index = text.find(substring, start)
        if index == -1:
            break
        positions.append(index)
        start = index + 1  # move forward to find overlapping matches

    return positions

# Example
text = "banana"
substring = "ana"
print(find_substring_occurrences(text, substring))
# Output: [1, 3]
```


###Q.6 Write a code to perform basic string compression using the counts of repeated characters.


```python
def compress_string(text):
    if not text:
        return ""

    compressed = []
    count = 1

    for i in range(1, len(text)):
        if text[i] == text[i - 1]:
            count += 1
        else:
            compressed.append(text[i - 1] + str(count))
            count = 1

    compressed.append(text[-1] + str(count))
    return "".join(compressed)

# Examples
print(compress_string("aaabcc"))       # a3b1c2
print(compress_string("aabcccccaaa"))  # a2b1c5a3
print(compress_string("abc"))          # a1b1c1
```

###Q.7 Write a code to determine if a string has all unique characters.


```python
def has_unique_characters(text):
    return len(set(text)) == len(text)

# Examples
print(has_unique_characters("python"))  # True
print(has_unique_characters("hello"))   # False
print(has_unique_characters(""))        # True
```

###Q.8 Write a code to convert a given string to uppercase or lowercase.


```python
def convert_case(text, case):
    if case == "upper":
        return text.upper()
    elif case == "lower":
        return text.lower()
    else:
        return "Invalid case option"

# Examples
print(convert_case("Hello World", "upper"))  # HELLO WORLD
print(convert_case("Hello World", "lower"))  # hello world
```

###Q.12 Implement a code to find the second largest number in a given list of integers.

```python
def find_second_largest(numbers):
    first = None
    second = None

    for num in numbers:
        if first is None or num > first:
            second = first
            first = num
        elif num != first and (second is None or num > second):
            second = num

    return second


# Example usage
nums = [12, 45, 7, 45, 31, 9]
print(find_second_largest(nums))  # Output: 31
```

###Q.14  Write a code to reverse a list in-place without using any built-in reverse functions.

```python
def reverse_list_in_place(lst):
    left = 0
    right = len(lst) - 1

    while left < right:
        # Swap elements at the two pointers
        lst[left], lst[right] = lst[right], lst[left]

        # Move toward the center
        left += 1
        right -= 1


# Example usage
numbers = [1, 2, 3, 4, 5]
reverse_list_in_place(numbers)

print(numbers)  # Output: [5, 4, 3, 2, 1]
```

###Q.15 Implement a code to find and remove duplicates from a list while preserving the original order of elements.

```python
def remove_duplicates(lst):
    seen = set()
    result = []

    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)

    return result


# Example usage
items = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
print(remove_duplicates(items))
# Output: [3, 1, 4, 5, 9, 2, 6]
```

###Q.16 Create a code to check if a given list is sorted (either in ascending or descending order) or not.


```python
def is_sorted(lst):
    ascending = all(lst[i] <= lst[i + 1] for i in range(len(lst) - 1))
    descending = all(lst[i] >= lst[i + 1] for i in range(len(lst) - 1))
    return ascending or descending


# Examples
print(is_sorted([1, 2, 3, 4]))   # True  (ascending)
print(is_sorted([9, 7, 5, 3]))   # True  (descending)
print(is_sorted([1, 3, 2]))      # False
print(is_sorted([5, 5, 5]))      # True  (both directions)
print(is_sorted([]))             # True
print(is_sorted([42]))           # True
```

###Q.17 Write a code to merge two sorted lists into a single sorted list.


```python
def merge_sorted_lists(list1, list2):
    merged = []
    i = j = 0

    # Compare elements and add the smaller one
    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            merged.append(list1[i])
            i += 1
        else:
            merged.append(list2[j])
            j += 1

    # Add any remaining elements
    merged.extend(list1[i:])
    merged.extend(list2[j:])

    return merged


# Examples
print(merge_sorted_lists([1, 3, 5], [2, 4, 6]))
# Output: [1, 2, 3, 4, 5, 6]

print(merge_sorted_lists([], [1, 2, 3]))
# Output: [1, 2, 3]

print(merge_sorted_lists([1, 1, 2], [1, 3]))
# Output: [1, 1, 1, 2, 3]
```

###Q.18 Implement a code to find the intersection of two given lists.


```python
def intersection(list1, list2):
    """Return the unique elements present in both lists."""
    return list(set(list1) & set(list2))


def intersection_with_duplicates(list1, list2):
    """Return common elements, respecting available duplicates."""
    counts = {}
    for item in list2:
        counts[item] = counts.get(item, 0) + 1

    result = []
    for item in list1:
        if counts.get(item, 0) > 0:
            result.append(item)
            counts[item] -= 1

    return result


# Examples
print(intersection([1, 2, 2, 3], [2, 3, 4]))
# Output: [2, 3]

print(intersection_with_duplicates([1, 2, 2, 3], [2, 2, 3, 4]))
# Output: [2, 2, 3]
```

###Q.19 Create a code to find the union of two lists without duplicates.


```python
def union_without_duplicates(list1, list2):
    """Return all unique elements from both lists, preserving first-appearance order."""
    return list(dict.fromkeys(list1 + list2))


def union_sorted(list1, list2):
    """Return all unique elements from both lists in ascending order."""
    return sorted(set(list1) | set(list2))


# Examples
print(union_without_duplicates([1, 2, 2, 3], [3, 4, 4, 5]))
# Output: [1, 2, 3, 4, 5]

print(union_sorted([1, 2, 2, 3], [3, 4, 4, 5]))
# Output: [1, 2, 3, 4, 5]
```

###Q.20 Write a code to shuffle a given list randomly without using any built-in shuffle functions.


```python
import random

def shuffle_list(lst):
    """Return a new randomly shuffled copy of the input list."""
    result = lst[:]  # Copy so the original list is unchanged

    # Start from the last index and swap with a random earlier index
    for i in range(len(result) - 1, 0, -1):
        j = random.randint(0, i)
        result[i], result[j] = result[j], result[i]

    return result


# Example
numbers = [1, 2, 3, 4, 5]
print(shuffle_list(numbers))
# Example output: [3, 1, 5, 2, 4]

print(numbers)
# Original remains: [1, 2, 3, 4, 5]
```

###Q.21 Write a code that takes two tuples as input and returns a new tuple containing elements that are Common to both input tuples.


```python
def common_elements(tuple1, tuple2):
    """Return a tuple containing elements present in both input tuples."""
    return tuple(dict.fromkeys(item for item in tuple1 if item in tuple2))


# Examples
print(common_elements((1, 2, 2, 3), (2, 3, 4)))
# Output: (2, 3)

print(common_elements(('a', 'b', 'c'), ('b', 'c', 'd')))
# Output: ('b', 'c')

print(common_elements((1, 2), (3, 4)))
# Output: ()
```

###Q.32 Write a code that takes two dictionaries as input and merges them into a single dictionary. If there are common keys, the values should be added together.


```python
def merge_dictionaries(dict1, dict2):
    merged = dict1.copy()

    for key, value in dict2.items():
        if key in merged:
            merged[key] += value
        else:
            merged[key] = value

    return merged


dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 5, "c": 15, "d": 25}

result = merge_dictionaries(dict1, dict2)

print("Dictionary 1:", dict1)
print("Dictionary 2:", dict2)
print("Merged dictionary:", result)
```

###Q.33 Write a code to access a value in a nested dictionary. The function should take the dictionary and a list of keys as input, and return the corresponding value. If any of the keys do not exist in the dictionary, the function should return None.


```python
def sort_dictionary_by_values(dictionary, ascending=True):
    return dict(sorted(dictionary.items(), key=lambda item: item[1], reverse=not ascending))


data = {"apple": 5, "banana": 2, "cherry": 8, "date": 1}

print("Ascending order:")
print(sort_dictionary_by_values(data, ascending=True))

print("\nDescending order:")
print(sort_dictionary_by_values(data, ascending=False))
```

###Q.34 Write a code that takes a dictionary as input and returns a sorted version of it based on the values. You can choose whether to sort in ascending or descending order.


```python
def invert_dictionary(dictionary):
    inverted = {}

    for key, value in dictionary.items():
        if value in inverted:
            inverted[value].append(key)
        else:
            inverted[value] = [key]

    return inverted


original = {
    "a": 1,
    "b": 2,
    "c": 1,
    "d": 3,
    "e": 2
}

inverted = invert_dictionary(original)

print("Original dictionary:", original)
print("Inverted dictionary:", inverted)
```
